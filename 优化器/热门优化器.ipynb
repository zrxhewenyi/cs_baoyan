```python
class SGD:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr

    def step(self, dx):
        self.x = self.x  - self.lr * dx 
```

```python
class Momentum:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr
        self.v = np.zeros_like(x)

    def step(self, dx, beta=0.9):
        self.v = beta * self.v + (1 - beta) * dx

        self.x = self.x  - self.lr * self.v
```

```python
# 有人叫做 牛顿冲量法 但是和牛顿法完全没有关系

# dx_v 是 $$ \frac{\partial(Loss_function(x + v))}{\partialx} $$

class Nesterov:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr
        self.v = np.zeros_like(x)

    def step(self, dx_v, beta=0.9):
        self.v = beta * self.v + (1 - beta) * dx_v

        self.x = self.x  - self.lr * self.v
```

```python
class RMSporp:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr
        self.s = np.zeros_like(x)

    def step(self, dx, beta=0.9):
        self.s = beta * self.s + (1 - beta) * (dx ** 2)

        self.x = self.x  - (self. lr / (np.sqrt(self.s) + 1e-6)) * dx
```

```python
class Adam:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr
        self.v = np.zeros_like(x)
        self.s = np.zeros_like(x)
        self.t = 0

    def step(self, dx, alpha=0.9, beta=0.9):
        self.t += 1

        self.v = alpha * self.v + (1 - alpha) * dx
        self.v = self.v / (1 - alpha ** self.t)

        self.s = beta * self.s + (1 - beta) * (dx ** 2)
        self.s = self.s / (1 - beta ** self.t)

        self.x = self.x  - (self. lr / (np.sqrt(self.s) + 1e-6)) * self.v
```

```python
# 与Adam的区别在于：把正则化项放在了优化器的内部，避免了很多问题(数学推导看笔记)

class AdamW:
    def __init__(self, x, lr=0.01):
        self.x = x
        self.lr = lr
        self.v = np.zeros_like(x)
        self.s = np.zeros_like(x)
        self.t = 0

    def step(self, dx, alpha=0.9, beta=0.9, Lambda=0.05):
        self.t += 1

        self.v = alpha * self.v + (1 - alpha) * dx
        self.v = self.v / (1 - alpha ** self.t)

        self.s = beta * self.s + (1 - beta) * (dx ** 2)
        self.s = self.s / (1 - beta ** self.t)

        self.x = (1 - Lambda) * self.x  - (self. lr / (np.sqrt(self.s) + 1e-6)) * self.v
```
